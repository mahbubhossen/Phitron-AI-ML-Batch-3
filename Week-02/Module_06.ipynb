# Different types of argument

```python
def adder(a,b,c):
    print(f"{a},{b},{c}")

adder(10,20,30)
```

```python
def adder(*args):
    print(args)


adder(10,20,30,40,50)
```

```python
def adder(*args):
    for i in args:
        print(i)


adder(10,20,30,40,50)
```

# keyword argument

```python
def adder(a,b,c):
    print(f"{a},{b},{c}")

adder(a=10,b=20,c=30)
```

```python
def adder(**kwards):
    print(kwards)


adder(a = 10 , b = 20, c = 40 , n_estimator = 50)
```

```python
def adder(**kwards):
    for key,value in kwards.items():
        print(f"{key}: {value}")


adder(a = 10 , b = 20, c = 40 , n_estimator = 50)
```

# Generator

```python
lst = [x for x in range (500)] # dataset

def data_loader (chunk_size, lst):
    for i in range (0,len(lst),chunk_size):
        yield lst[i:i+chunk_size]

x = data_loader(5,lst)
print(type(x))

print(next(x))
print(next(x))
```

# lambda , map , apply function

```python
import pandas as pd

df = pd.DataFrame({'temperature (F)': [32, 212, 98.6, 68]})
df
```

```python
df['temperature (C)'] = df['temperature (F)'].apply(lambda x: (x-32)*5/9)
print(df)
```

```python
import pandas as pd

df = pd.DataFrame({'A': [1,2,3,4], 'B':[5,6,7,8]})
df
```

```python
df['A'] =  df['A'].map(lambda x: x**0.5)
df
```

```python
import pandas as pd

df = pd.DataFrame({'total_spent': [120,600,450,800]})

# creating a high value feature on the fly
df['is_high_value'] = df['total_spent'].apply(lambda x: 1 if x>500 else 0)
df
```

# F. GPU Can Eat Only Three Images (Hard)

```python
def batch_generator(data, batch_size):
    if batch_size<1:
        return

    for i in range(0,len(data),batch_size):
        yield data[i:i+batch_size]

def batch_mean(batch):
    return sum(batch)/len(batch)

data = [1,2,3,4,5,6,7,8,9]
batch_size = 3

for batch in batch_generator(data,batch_size):
    print(f"{batch} mean={batch_mean(batch)}")
```
