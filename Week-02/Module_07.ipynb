# Practice 1:

Get the text from sample.text, and count the following

Number of lines

Number of words

Number of characters

and save it to the counter_of_string.txt file

```python
from functools import reduce

with open("sample.txt", "r") as file:
    strings = file.readlines()

    # total line
    total_lines = len(strings)
    print(total_lines)

    # total words
    number_of_words = list(map(lambda x: len(x.split()), strings))
    print(number_of_words)

    # total number of words
    total_number_of_words = reduce(lambda x,y:x+y, number_of_words)
    print(total_number_of_words)


    # cleaning process

    # new line delete
    strings = list(map(str.strip, strings))
    print(strings)

    #space delete
    strings = list(map(lambda x: x.replace(" ",""), strings))
    print(strings)


    # number of character ber korbo
    number_of_character = list(map(lambda x: len(x), strings))
    print(number_of_character)

    total_number_of_character = reduce(lambda x,y:x+y, number_of_character)
    print(total_number_of_character )


# save file

with open("counter_of_string.txt","w") as file:
    file.write(f"total line: {total_lines}\ntotal number of words :{total_number_of_words}\ntotal number of characters: {total_number_of_character}")
```
