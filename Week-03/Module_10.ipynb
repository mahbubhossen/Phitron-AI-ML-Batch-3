# Nd Array

```python
# 1d array

arr1 = np.array([1,2,3,4,5])
# print(type(arr1)


# 2d array

arr2 = np.array([[1,2,3],[4,5,6]])


# 3d array
arr3 = np.array([
    #1st floor
    [[1,2,3],[4,5,6]],
    # 2nd floor
    [[1,2,3],[4,5,6]]
])
```

# array attributes

```python
# dimantion

print(arr1.ndim)
print(arr2.ndim)
print(arr3.ndim)


# shape

print(arr1.shape)
print(arr2.shape)
print(arr3.shape)


# data type
print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)


# size
print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
# dictionary

dc = {'a':10, 'b':20, 'c':30}

print(type(dc))

keys = dc.keys()
print(keys)

values = dc.values()
print(values)

items = dc.items()
print(items)


# keys to ndarray
arr = np.array(list(keys))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# values to ndarray
arr = np.array(list(values))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# items to ndarray
arr = np.array(list(items))
print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# Array creation with range functions

```python
# np.arange(first,end,step) -> similar to range function

arr = np.arange(1,10,1)
mat = arr.reshape(3,3)

print(arr)
print(mat)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.linspace(start,end(included),koita)

arr = np.linspace(1,10,15)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# np.logspace(start,end(included),koita)

arr = np.logspace(0,4,6, base=2)

print(arr)
print(type(arr))
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

# Indexing and slicing

```python
# indexing

print(arr1)
print(arr2)

# print(arr1[2])

arr1[2] = 100

print(arr1)


# print(arr2[0][1])
arr2[0][1] = 100
print(arr2)
```

```python
# slicing

# 1d = arr[start:end:step]

arr1_mod = arr1[1:4]
print(arr1_mod)
```

```python
# 2d slicing

# arr [row_start: row_end : row_step , column_start: column_end: column_step]

print(arr2)

# getting a row

row_0 = arr2[0:1,]
# print(row_0)

row_1 = arr2[1: ,]
# print(row_1)


# getting  a column

col_0 = arr2[:: , 0:1]
print(col_0)

col_1 = arr2[:: , 1:2]
print(col_1)


# getting a portion
portion = arr2[::, 0:2]
print(portion)
```

```python
# copy

arr1_mod = arr1[1:4].copy()
arr1[1] = 2

arr1_mod[2] = 200
print(arr1_mod)
```

# Extra Practice Problem

```python
"""
1. You have a list of temperatures recorded in Celsius. Your task is to:
Create a NumPy array from the given list.
Convert all temperatures from Celsius to Fahrenheit. The formula is F = C* 1.8 + 32.
Print the new Fahrenheit array.
"""


import numpy as np

celsius = [0,10,20,30,40]

celsius_array = np.array(celsius)

fahrenheit = celsius_array * 1.8 + 32


print(fahrenheit)
```

```python
"""
2.  You have a Python list containing the lap times (in seconds) for an athlete.

Task:

Convert the list into a NumPy array.
Print the array's shape, size, and data type.
The athlete wants to see their times in minutes.
Create a new array by dividing all the lap times by 60.
Print the new array of lap times in minutes.

"""



import numpy as np

lap_times = [120,150,180,210,240]

lap_array = np.array(lap_times)


print("shape: " , lap_array.shape)
print("size: ", lap_array.size)
print("Data type: ", lap_array.dtype)

minutes = lap_array/60

print("lap times in minutes : ", minutes)
```
