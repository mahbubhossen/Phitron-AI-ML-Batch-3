# String Methods

```python
string = "Welcome to Phitron ML Course , Phitron ML , Hello ML"

processed_string = string.lower()
print(processed_string)

# Length
print(len(string))

# word khuja
print("phitron" in processed_string)


# surur kon index theke substring ta suru hocche
start_index = processed_string.find("phitron")
print(start_index)



# sesher kon index theke substring ta suru hocche
end_index = processed_string.rfind("phitron")
print(end_index)


#substring koybar ache
count = processed_string.count("phitron")
print(count)



#replace a substring
new_string = string.replace("ML", "AL/ML")
print(new_string)
```

# list comprehension

```python
even = []

for i in range(101):
    if i%2 == 0:
        even.append(i)
print(even)
```

```python
random_list = [x for x in range(10)]
print(random_list)

random_list2 = [x*2 for x in range(10)]
print(random_list2)

random_list3 = [x for x in range(10) if x%2 == 0]
print(random_list3)
```

```python
fruits = ["apple", "banana", "mango", "orange"]

capitalize_fruits = []

for fruit in fruits:
    print(fruit)
    capitalize_fruits.append(fruit.capitalize())

print(capitalize_fruits)
```

```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
lowercase_fruits = [fruit.lower() for fruit in fruits]
print(lowercase_fruits)
```
