
# Problem 4: Count Even Numbers

```python
t = int(input())

n = list(map(int, input().split()))

count = 0
for i in range(t):

    if n[i] % 2 == 0:
        count+=1
print(count)
```

# Problem 5: Sum of Odd Numbers

```python
t = int(input())

sum = 0

for i in range(t):
    if i % 2 == 0:
        continue
    sum+=i
print(sum)
```

# Problem 6: Smallest and Largest

```python
t = int(input())

n = list(map(int , input().split()))

minimum = n[0]
maximum = n[0]
for i in range (t):
    if minimum > n[i]:
        minimum = n[i]
    elif maximum < n[i]:
        maximum = n[i]
print(minimum , maximum)

```

# Problem 7: Positive, Negative, Even and Odd

```python
t = int(input())

n = list(map(int , input().split()))

positive = 0
negative = 0
even = 0
odd = 0

for i in range(t):
    if n[i] > 0:
        positive += 1
    if n[i] < 0:
        negative += 1
    if n[i] % 2 == 0:
        even += 1
    if n[i] % 2 != 0:
        odd += 1


print("Positive:",positive)
print("Negative:",negative)
print("Even:",even)
print("Odd:",odd)
```

# Problem 8: Reverse Digits

```python
n = int(input())

while n>0:
    digit = n%10
    print(digit, end=" ")
    n = n//10
```

# Problem 9: Find the First Number Greater Than 100

```python
t = int(input())
n = list(map(int , input().split()))

for i in range(t):
    if n[i]>100:
        print(n[i])
        break
```

# Problem 10: Number Analysis

```python
t = int(input())

n = list(map(int , input().split()))\


count = 0

for i in range(t):
    if n[i] % 2 == 0:
        continue

    num = n[i]
    digit_sum = 0

    while num>0:
        digit_sum += num % 10
        num = num // 10

    if digit_sum > 20:
        break



    print(n[i], digit_sum)
    count += 1


print(count)
```
