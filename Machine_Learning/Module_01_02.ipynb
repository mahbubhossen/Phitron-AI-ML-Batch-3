# univariate analysis (categorical )

```python
sns.countplot(data = df, x = df['Survived'])
```

```python
df['Survived'].value_counts()
```

```python
(df['Survived'].value_counts() / len(df))*100
```

```python
survived_count = df['Survived'].value_counts()

survived_label = df['Survived'].unique()
print(survived_label)


plt.pie(survived_count , labels = survived_label,  autopct="%1.1f")
plt.title('Pie chart of survival rate ')
plt.show()
```

```python
sns.countplot(data = df, x = df['Pclass'])
```

```python
(df['Pclass'].value_counts() / len(df))*100
```

```python
sns.countplot(data = df, x = df['Sex'])
```

```python
(df['Sex'].value_counts())/len(df) * 100
```

```python
sns.countplot(data = df, x = df['Embarked'])
```

```python
(df['Embarked'].value_counts())/len(df) * 100
```

```python
sns.histplot(df['Age'] , bins=50)

plt.title('Histrogram of Age')
plt.xlabel('Ages')
plt.ylabel('frequency')

plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('KDE plot of Ages')
plt.xlabel('Ages')
plt.ylabel('probality distribution')

plt.show()
```

```python
sns.histplot(df['Fare'] , bins=50)

plt.title('Histrogram of Fare')
plt.xlabel('Fare')
plt.ylabel('frequency')

plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title('KDE plot of Fare')
plt.xlabel('Fare')
plt.ylabel('probality distribution')

plt.show()
```

```python
sns.boxplot(df['Fare'])
```

# multi variate analysis

```python
sns.countplot(x=df['Sex'], hue=df['Survived'])
```

```python
gender_split = df.groupby(['Sex'])['Survived'].value_counts()
print(gender_split)
```

```python
gender_split = df.groupby(['Sex'])['Survived'].value_counts(normalize=True) * 100
print(gender_split)
```

```python
sns.countplot(x=df['Pclass'], hue=df['Survived'])
```

```python
pclass_split = df.groupby(['Pclass'])['Survived'].value_counts()
print(pclass_split)
```

```python
pclass_split = df.groupby(['Pclass'])['Survived'].value_counts(normalize = True) * 100
print(pclass_split)
```
