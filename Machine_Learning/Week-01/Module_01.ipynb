# univariate analysis (categorical )

```python
sns.countplot(data = df, x = df['Survived'])
```

```python
df['Survived'].value_counts()
```

```python
(df['Survived'].value_counts() / len(df))*100
```

```python
survived_count = df['Survived'].value_counts()

survived_label = df['Survived'].unique()
print(survived_label)


plt.pie(survived_count , labels = survived_label,  autopct="%1.1f")
plt.title('Pie chart of survival rate ')
plt.show()
```

```python
sns.histplot(df['Age'] , bins=50)

plt.title('Histrogram of Age')
plt.xlabel('Ages')
plt.ylabel('frequency')

plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('KDE plot of Ages')
plt.xlabel('Ages')
plt.ylabel('probality distribution')

plt.show()
```

```python
sns.histplot(df['Fare'] , bins=50)

plt.title('Histrogram of Fare')
plt.xlabel('Fare')
plt.ylabel('frequency')

plt.show()
```

```python
sns.kdeplot(df['Fare'])
plt.title('KDE plot of Fare')
plt.xlabel('Fare')
plt.ylabel('probality distribution')

plt.show()
```
